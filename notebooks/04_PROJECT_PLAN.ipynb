# St. Paul Neighborhood Health Project## Research Framework & Strategic Plan

## Project OverviewThis project analyzes neighborhood health in St. Paul using four key datasets:- **Building Permits** - Development & housing investment- **Crime Incidents** - Public safety patterns- **Housing Production** - New housing developments- **Service Requests** - Resident-reported issues and needs### Research Question**How do building development, public safety, housing production, and service request patterns vary across St. Paul neighborhoods, and what do these patterns reveal about neighborhood health?**

## Understanding Each Dataset### Building Permits (~130MB)**What it tells us:**- Level of development and investment in neighborhoods- Types of construction (residential, commercial, mixed-use)- Timeline and velocity of development- Economic activity and neighborhood trajectory**Key analyses:**- Permits per neighborhood over time- Mix of residential vs. commercial permits- Permit value/investment by area- Development hotspots vs. stagnant areas

### Crime Incident Reports (~82MB)**What it tells us:**- Public safety and crime patterns- Types of crimes in different areas- Seasonal and temporal trends- Neighborhood safety levels**Key analyses:**- Crime rate per neighborhood (absolute and per capita)- Crime type distribution by neighborhood- Violent vs. non-violent patterns- Temporal trends (improving or worsening?)

### Housing Production GeoJSON (~0.8MB)**What it tells us:**- Actual new housing units created- Geographic distribution of developments- Housing types and sizes- Development density and clustering**Key analyses:**- Units per neighborhood per year- Housing type distribution- Geographic clustering- Alignment with permit data

### Resident Service Requests (~57MB)**What it tells us:**- Issues residents care about and report- Problem frequency by neighborhood- Infrastructure and maintenance needs- Quality of life indicators**Key analyses:**- Top issue types by neighborhood- Request volume and trends- Response/resolution times (if available)- Issue severity patterns

## Health Indicators### Safety IndexMeasures public safety and crime levels:- **Components:** Violent crime rate, property crime rate, trend- **Scale:** 0-100 (higher is safer)- **Benchmark:** City average = 50 points### Development IndexMeasures housing and economic development:- **Components:** Permits per capita, housing units, investment- **Scale:** 0-100 (higher is more developed)- **Interpretation:** Active vs. stagnant markets### Service Need IndexMeasures resident-reported problems:- **Components:** Request volume per capita, resolution time- **Scale:** 0-100 (higher = more requests/need)- **Interpretation:** Infrastructure burden and quality of life### Composite Health Score**Formula:**```Health_Score = (Safety_Index × 0.40) +                (Development_Index × 0.30) +                (Service_Need_Index × -0.30)```- **Range:** 0-100- **70+:** Healthy neighborhood- **50-70:** Moderate health- **<50:** Needs attention

## Key Research Questions

### Primary Questions1. **What is the health profile of each neighborhood?**   - Safety levels, development activity, service needs   - Ranking neighborhoods by composite health score2. **How do neighborhoods cluster by characteristics?**   - Similar patterns and profiles   - Grouping neighborhoods for policy3. **Which neighborhoods are improving vs. declining?**   - Trajectory analysis across time   - Identifying trends

### Secondary Questions4. Does more development correlate with crime changes?5. Are safer neighborhoods developing faster?6. Where are service requests concentrated?7. What types of problems are most common by area?8. Are there geographic patterns or hotspots?9. How have neighborhoods changed over the analysis period?
