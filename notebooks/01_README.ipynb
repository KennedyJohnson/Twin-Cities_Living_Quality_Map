### Step 1: Explore Your Data
Run the exploratory data analysis to understand dataset structure.

````python
import pandas as pd
import numpy as np

# Load datasets (adjust paths as needed)
perms = pd.read_csv('../data/Approved_Building_Permits_-7890413957898939046.csv')
crime = pd.read_csv('../data/Crime_Incident_Report.csv')
requests = pd.read_csv('../data/Resident_Service_Requests_7024824928576740068.csv')

print("✓ Datasets loaded successfully")
print(f"Permits shape: {perms.shape}")
print(f"Crime shape: {crime.shape}")
print(f"Requests shape: {requests.shape}")
````

### Step 2: Understand Your Data Structure

````python
# Check column names
print("Permit columns:")
print(perms.columns.tolist())
print("\nCrime columns:")
print(crime.columns.tolist())
print("\nRequest columns:")
print(requests.columns.tolist())
````

### Step 3: Identify Neighborhood Columns

````python
# Find neighborhood/geographic identifiers
for df, name in [(perms, 'Permits'), (crime, 'Crime'), (requests, 'Requests')]:
    print(f"\n{name} data:")
    for col in df.columns:
        if any(term in col.lower() for term in ['neighbor', 'district', 'ward', 'zone', 'area']):
            print(f"  Found: '{col}' with {df[col].nunique()} unique values")
            print(f"    Sample: {df[col].unique()[:3]}")
            print(f"    Data type: {df[col].dtype}")
````
