# Exploratory Data Analysis (EDA)
## St. Paul Neighborhood Health Project

This notebook loads and explores all four datasets to understand structure, quality, and key features.

## Setup

Load required libraries and datasets.

```python
import pandas as pd
import numpy as np
import json
import warnings
warnings.filterwarnings('ignore')

print('Libraries loaded successfully')
```

```python
# Load datasets
print("Loading datasets...\n")

try:
    permits = pd.read_csv('../data/Approved_Building_Permits_-7890413957898939046.csv')
    print(f"✓ Building Permits loaded: {permits.shape[0]:,} rows, {permits.shape[1]} columns")
except FileNotFoundError:
    print("✗ Building Permits not found. Check filename and path.")

try:
    crime = pd.read_csv('../data/Crime_Incident_Report.csv')
    print(f"✓ Crime Incidents loaded: {crime.shape[0]:,} rows, {crime.shape[1]} columns")
except FileNotFoundError:
    print("✗ Crime Incidents not found. Check filename and path.")

try:
    requests = pd.read_csv('../data/Resident_Service_Requests_7024824928576740068.csv')
    print(f"✓ Service Requests loaded: {requests.shape[0]:,} rows, {requests.shape[1]} columns")
except FileNotFoundError:
    print("✗ Service Requests not found. Check filename and path.")

try:
    housing = pd.read_csv('../data/Housing_Production.csv')
    print(f"✓ Housing Production loaded: {housing.shape[0]:,} rows, {housing.shape[1]} columns")
except FileNotFoundError:
    print("✗ Housing Production not found. Check filename and path.")

print("\n" + "="*80)
```

## Building Permits Analysis

```python
print("BUILDING PERMITS DATASET")
print("="*80)
print(f"\nShape: {permits.shape}")
print(f"\nColumns ({permits.shape[1]}):")
for col in permits.columns:
    print(f"  - {col}")
print(f"\nData Types:")
print(permits.dtypes)
```

```python
print("\nFirst Few Rows:")
print(permits.head(2))
```

```python
print("\nMissing Values:")
missing = permits.isnull().sum()
missing_pct = (missing / len(permits) * 100).round(1)
missing_df = pd.DataFrame({
    'Column': missing.index,
    'Missing Count': missing.values,
    'Missing %': missing_pct.values
}).sort_values('Missing %', ascending=False)
print(missing_df[missing_df['Missing Count'] > 0])
```

```python
print("\nBasic Statistics:")
print(permits.describe())
```

```python
# Look for neighborhood/geographic columns
print("\nGeographic/Neighborhood Columns:")
for col in permits.columns:
    if any(term in col.lower() for term in ['neighbor', 'district', 'ward', 'zone', 'area', 'location', 'geo']):
        print(f"\n'{col}':")
        print(f"  Type: {permits[col].dtype}")
        print(f"  Unique values: {permits[col].nunique()}")
        print(f"  Samples: {permits[col].unique()[:5]}")
        print(f"  Missing: {permits[col].isnull().sum()}")
        if permits[col].nunique() <= 20:
            print(f"  Value counts:\n{permits[col].value_counts()}")
        else:
            print(f"  Top 10:\n{permits[col].value_counts().head(10)}")
        break
```

## Housing Production (GeoJSON) Analysis

```python
print("HOUSING PRODUCTION (GeoJSON)")
print("="*80)
print(f"\nType: {type(housing)}")
if isinstance(housing, dict):
    print(f"Keys: {list(housing.keys())}")
    if 'features' in housing:
        print(f"Number of features: {len(housing['features'])}")
        print(f"\nSample feature (first 500 chars):")
        import json as json_module
        sample = json_module.dumps(housing['features'][0], indent=2)[:500]
        print(sample)
```

```python
# Extract properties from GeoJSON features
if isinstance(housing, dict) and 'features' in housing:
    feature_properties = []
    for feature in housing['features']:
        if 'properties' in feature:
            feature_properties.append(feature['properties'])
    
    housing_df = pd.DataFrame(feature_properties)
    print(f"\nHousing Production Data (from GeoJSON):")
    print(f"Shape: {housing_df.shape}")
    print(f"\nColumns: {housing_df.columns.tolist()}")
    print(f"\nData types:")
    print(housing_df.dtypes)
    print(f"\nFirst few rows:")
    print(housing_df.head(2))
    print(f"\nBasic statistics:")
    print(housing_df.describe())
```

```python
# Create summary table
summary_data = {
    'Dataset': ['Building Permits', 'Crime Incidents', 'Service Requests', 'Housing Production'],
    'Rows': [len(permits), len(crime), len(requests), len(housing_df) if 'housing_df' in locals() else len(housing.get('features', []))],
    'Columns': [permits.shape[1], crime.shape[1], requests.shape[1], len(housing_df.columns) if 'housing_df' in locals() else 'N/A'],
    'Max Missing %': [
        (permits.isnull().sum() / len(permits) * 100).max(),
        (crime.isnull().sum() / len(crime) * 100).max(),
        (requests.isnull().sum() / len(requests) * 100).max(),
        'N/A'
    ]
}

summary_df = pd.DataFrame(summary_data)
print("\nDATA QUALITY SUMMARY:")
print(summary_df.to_string(index=False))
```
